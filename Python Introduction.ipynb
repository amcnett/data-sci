### A Few Main Differences Between Python and Java

1. No data type declaration - dynamically typed language
2. Semi-colons optional at end of statements
3. Strings can be either single or double quoted (no chars)

```python
cost = 10  # numeric type int
tax = .06  # numeric type float
total = cost + cost * tax
print(f"Total is {total:.2f}")  # prints a string

print(type(cost))
cost = float(cost) # casting
print(type(cost))

# later you'll see sequence data types (e.g. lists) and mapping types (e.g. dictionary)
```

4. Lose the curly braces, indentation is required in blocks (be consistent in spaces for indentation)
5. If statements are written as if, elif, else with colon at end of expression

```python
print (cost > 20)  # prints bool data type

if cost > 20:  # note the quote at the end of the expression!
    tax = .07
elif cost > 15:
    tax = .06
else:
    tax = 0.0
    
total = cost + cost * tax
print(f"Total is {total:.2f}")
```

6. Function syntax is a bit different

```python
def calc_total(cost):  # not use of keyword "def"
    if cost > 20:
        tax = .07
    elif cost > 15:
        tax = .06
    else:
        tax = 0.0
    return cost + cost * tax

print("Total is: ", calc_total(20))
```

Create a function that determines if a triangle is a triangle (any 2 sides must be greater than the third)

Try the following:

1, 1, 1 - triangle

1, 2, 1 - not a triangle

1, 2, 3 - not a triangle

2, 2, 3 - triangle

```python
def isTriangle(side1, side2, side3):
    if side1 + side2 > side3 and side2 + side3 > side1 and side1 + side3 > side2:
        return True
    return False


print(isTriangle(1,1,1))
print(isTriangle(1,2,1))
print(isTriangle(1,2,3))
print(isTriangle(2,2,3))
```

7. For loop is written more like a foreach loop

```python
# loop to print multiples of 7

n = 0
while n < 50:
    if(n % 7 == 0):
        print (n)
    n += 1
```

```python
# Loop to print multiples of 7 (for loop equivalent)
for n in range(0, 50):
    if(n % 7 == 0):
        print (n)
```

```python
# Loop to print multiples of 7 (for loop equivalent)
for n in range(50):
    if(n % 7 == 0):
        print (n)
```

8. No arrays. Instead we have lists, tuples, sets, and dictionaries.
See [list documentation]( https://docs.python.org/3/tutorial/datastructures.html).

```python
# lists are ordered, changeable (mutable),and can have duplicate values
scores = [10, 23, 34, 90, 23]  # a list store multiple items
print("length is", len(scores)) # number of items in scores
print("first score is", scores[0]) # print first item
scores[1] = 55 # update second item

for score in scores:
    print(score)
    
missed_scores = [68, 45]
scores = scores + missed_scores
print(scores)
```

One nice aspect of working with lists in Python is that you can do **slicing**. Slicing involves using the **:** operator.  Slicing gets you a range of values.

```python
college = "Pennsylvania College of Technology"

# get college
print(college[13:20])

# get Pennsylvania  -- slice from start
print(college[:12])

# get Technology -- slice to end
print(college[24:])

# negative indexing
# get Technology -- slice from end
print(college[-10:])

# skip letters/elements
print(college[::2])
```

```python
# tuples are ordered and can have duplicates, but cannot change (immutable)
fruits = ("orange", "banana", "pear", "orange")
# fruits[1] = "kiwi"  # this action is not supported
# fruits.add("kiwi") # this action is not supported
print(fruits)
```

```python
# named tuples can be indexed with strings
from collections import namedtuple

Veg = namedtuple("Veg", ['cost', 'qty'])
carrot = Veg(1.99, 100)
print(carrot.cost)
```

```python
# sets are unordered, unchangeable (but can add and remove items), and do not allow duplicate values
names = {"mario", "luigi", "peach", "yoshi"}
names = {"mario", "luigi", "peach", "yoshi", "mario"} # dups are ignored
print(names)
# print(names[0]) # not indexed, cannot do this!
names.add("wario")
names.remove("luigi")
print(names)
```

I have a list but I want to remove duplicates. How can I use a set to do this? Hint: to create a list I can use the list() function.  To create a set I can also use the set() function.
