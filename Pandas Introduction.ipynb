## Pandas Introduction

Pandas has two types of classes for handling data:
* series
* dataframe/DataFrame

**Series** are a one-dimensional labeled array while **DataFrames** are two-dimensional.

We import the pandas package often to aid us in manipulating datasets via the use of the DataFrame object type. Like tables, dataframes have rows and columns. Rows are typically labeled via an index while columns are given specific string labels. Again, like tables, columns must have the same data type, but different columns can contain different data types.  

**Dataframes are index, ordered, and mutable.** Unlike ndarrays, dataframes are always two-dimensional.  Dataframes also have the advantage of permitting us to access data via string labels instead of integer indexes only. 

Dataframes can be created several ways. You'll see a few examples in the code provided in this notebook.

```python
import numpy as np # so we can reference if we needed it - pandas is actually built on top of numpy
import pandas as pd # we typically import as pd

# Create dataframe with 7 rows, columns. 
# Fill with values from the standard normal distribution 
df2 = pd.DataFrame(np.random.randn(7, 3), columns=['A', 'B', 'C'])
df2
#df2.shape
```

Let's explore the column and row labels.

```python
print(df2.columns)
print(df2.index)
# note that .values will return an array of values
```

What about accessing values?  To access and entire column, we can use the dataframe_name['column name']. Mutliple columns require multiple square brackets.  

```python
temp = df2['A']
temp
```

Note that what is returned above is actually a **series**.  We can get the same thing by also stating df2.A.

```python
df2.A
```

If you want a dataframe back, you need to use [[]] notation instead.

```python
df2[['A']]
```

We can also use double square brackets to reference more than one column. 

```python
print(df2[['A','B']])
```

To acccess rows, we need to use the iloc() dataframe method. If we want to refer to items by column labels instead, we can use the the .loc() dataframe method.  

```python
df2.iloc[0] # returns the values of the first row
```

```python
df2.iloc[0:2] # returns the values of the first two rows
```

```python
df2.iloc[0, 2] # returns the value in the first row, 3rd column
```

We can use the iloc and loc to also update individual values.

```python
df2.iloc[0, 2] = 2 # updates the value in the first row, 3rd column
df2
```

We can also apply slicing to dataframes. Slicing allows you to select multiple rows and/or multiple columns.

```python
df2.iloc[0:2, 1:3]
```

```python
# can do the same thing
df2.loc[0:2,['B','C']]  # note the use of "loc" vs "iloc"
```

```python
# can do the same thing
df2.loc[:2,['B','C']]  
```

### Filtering data with comparison operators

We can use traditional relational operators to view only a subset of the data as well, much like we'd do with a query. The catch is that the comparison needs to happen between the square brackets.

Consider the two examples below:

```python
df2['A'] > 0
```

```python
# vs
df2[df2['A'] > 0]
```

We can combine our tests using & for and, | for or, and ~ for not.  Note these are NOT the bitwise operators, but instead an element-wise operator for use with pandas. 

```python
# note that the parens are need to enforce precedence
df2[(df2['A'] > 0) & (df2['C'] > 0)]
```

### Dataframe Methods

Just like there are several methods to use to manipulate ndarrays, pandas include several methods (https://pandas.pydata.org/docs/reference/frame.html) for manipulating dataframes. Common methods to consider include:

* drop() to remove rows or columns
* drop_duplicates() to remove duplicate data
* dropna() to remove missing values
* insert() to insert column in a dataframe
* replace() to replace values in a dataframe
* sort_values() to sort values in a dataframe

```python
# Create dataframe using a list of lists
df = pd.DataFrame([['Bob', 83, 45],
                   ['Ann', 86, 22],
                   ['Lisa', 100, 31],
                   ['J', 51, 22],
                   ['Max', 19, 26],
                   ['Max', 19, 26]])

# Assign column names
df.columns = ['name', 'credits', 'age']

df
```

Pandas provides for easy ways to drop values from the dataframe.

```python
# remove duplicate records
#df = df.drop_duplicates()
df.drop_duplicates(inplace=True)

df  # does this update the original dataset?
```

```python
df = df.drop_duplicates()
df
```

There are also easy ways to insert and modify values.

```python
df.insert(loc=0, column='id', value=[123, 234, 345, 567, 675])
df
```

```python
# Can pair with crazy list comprehensions
df.insert(loc=0, column='class', value=['U' if credits < 120 else "G" for credits in df['credits']])
df
```

```python
# Adding an empty values so that we can practice removing it
df.loc[0, 'credits'] = np.nan
df
```

It is important to pay attention to the variety of options when using a given method.  For example, consider dropna().(https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.dropna.html)

```python
df.dropna() # default is to remove entire row where missing value is found
```

```python
df.dropna(axis="columns", inplace=True) #drop column where at least one element is missing
```

```python
df.dropna(how="all", inplace=True) # drop rows where all elements are missing
```

```python
# define what columns to look in
df.dropna(subset=['id', 'credits'])
df
```
