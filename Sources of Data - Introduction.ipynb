## APIs

While we can get data using databases and data sets available on sites like Kaggle, we can also get live data from sites. We discussed scraping as one method. The other more reliable method is API use.  

API stands for Application Programming Interface. APIs are used to more easily enable software to communicate. 

Benefits of APIs:
* data is provided in real-time
* can access a lot of data
* data comes to you in a pre-processed format

Common challenges:
* many social media sites now charge a premium to use APIs
* rate limits; you can still be blocked if you abuse the service and don't consider etiquette

### RESTful APIs

APIs exist in different types. One of the popular types, which we will use here is the Representational State Transfer (REST) API. In  RESTful API call, certain constraints are to be followed to adhere to the RESTful style/constraints. 

Calls to APIs depend on 2 things: the endpoint and the operation. The endpoint is the URL for the resource you are accessing. The operation is what operation you want to perform. Common operations are HTTP methods including GET, POST, PUT, and DELETE. Some follow this basic design:

* create operations -> use POST
* read operations -> use GET
* update operations -> use PUT
* delete operations -> used DELETE

For this exercise, you should be familiar with GET requests. The GET request is used to access resources. The POST request is used to send data to a server. For example, we can use the GET request to view data for a given YouTube video posted. We can use a POST request to create a new YouTube video posting. 

Best practices: 
* read the documentation
* cache your requests to avoid unneccessary requests

Many sites have APIs that allow you to access their data more easily. In the past, there was often free tier access with limitations. However, over the recent year we have seen a trend where social media companies are now monetizing more actively their APIs and we are seeing increasing limitations on free API use. So we are going to use a public/open and free API instead.

Note concerning authentication: APIs can also often require an API key or OAuth token that accompanies your request. To make this easier to follow along and start simple, we are going to use some public APIs that don't require authentication.
